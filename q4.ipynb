```python
def reverse_string(s):
    """
    Reverses a given string.

    Args:
        s (str): The input string to be reversed.

    Returns:
        str: The reversed string.

    Raises:
        TypeError: If the input 's' is not a string.
    """
    if not isinstance(s, str):
        raise TypeError("Input must be a string.")
    return s[::-1]

# Example usage:
original_string = "Hello World"
reversed_str = reverse_string(original_string)
print(f"Original string: {original_string}")
print(f"Reversed string: {reversed_str}")

original_string_2 = "Python"
reversed_str_2 = reverse_string(original_string_2)
print(f"Original string: {original_string_2}")
print(f"Reversed string: {reversed_str_2}")
```
