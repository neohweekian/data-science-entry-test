```python
def swap(x, y):
    # Check if x and y are numeric
    if not isinstance(x, (int, float)) or not isinstance(y, (int, float)):
        #print("Both x and y must be numeric values.")
        return print(f"{-1} : Both x and y must be numeric values")

    # Swap values using only x and y as variables
    x, y = y, x

    # Print the swapped values
    print(f"Swapped values: {x}, {y}")

#Example usage
swap("Apple",10) #Output is "Both x and y must be numeric values."
swap(9,17)       #Output is: Swapped values: 17,9
```
