```python
def check_divisibility(num, divisor):
    """
    Checks if a number (num) is divisible by another number (divisor).
    Both num and divisor must be numeric.

    Args:
        num (numeric): The number to be checked for divisibility.
        divisor (numeric): The divisor.

    Returns:
        bool: True if num is divisible by divisor, False otherwise.
    """
    if not isinstance(num, (int, float)) or not isinstance(divisor, (int, float)):
        print("Error: Both num and divisor must be numeric.")
        return False

    if divisor == 0:
        print("Error: Divisor cannot be zero.")
        return False

    return num % divisor == 0

# Scenario 1: num=10, divisor=2
result1 = check_divisibility(10, 2)
print(f"Is 10 divisible by 2? {result1}")

# Scenario 2: num=7, divisor=3
result2 = check_divisibility(7, 3)
print(f"Is 7 divisible by 3? {result2}")
```
