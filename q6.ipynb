```python
def find_first_negative(lst):
    """ Return the first negative number in a list if found,
    otherwise return 'No negatives'. Use a while loop to implement this.
    index of a list starts with zero, thus i < len(lst) """
    i = 0
    while i < len(lst):
        if lst[i] < 0:
            return lst[i]
        i += 1
    return "No negatives"

# Scenario 1
list1 = [3, 5, -1, 7, -2, 8]
print(f"Scenario 1: List {list1}, First negative: {find_first_negative(list1)}")

# Scenario 2
list2 = [2, 10, 7, 0]
print(f"Scenario 2: List {list2}, First negative: {find_first_negative(list2)}")
```
