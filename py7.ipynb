```python
#Define a class named Car with attributes: make, model, year
class Car:
    #Initialize the class with make, model, and year attributes
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    """ method named describe_car() that prints information about the car as
    'Year Make Model'"""
    def describe_car(self):
        print("My car information is:")
        print(f"Year={self.year} Make={self.make} Model={self.model}")

# Create an instance of the Car class
my_car = Car(make="Toyota", model="Corolla", year=2020)

# Call the describe_car method
my_car.describe_car()
```
