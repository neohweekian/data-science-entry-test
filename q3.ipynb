```python
def update_dictionary(dct, key, value):
    """
    Updates a dictionary with a new key-value pair. If the key already exists,
    print the original value before updating. Returns the updated dictionary.
    """
    if key in dct:
        print(f"Key '{key}' already exists. Original value: {dct[key]}")
    dct[key] = value
    return print(f"Updated dictionary: {dct}")
```

Below is to test the `update_dictionary` function with a few data points before the required scenarios:

```python
from typing_extensions import Dict
# Initialize a dictionary
dct = {'apple': 1, 'banana': 2}
print(f"Initial dictionary: {dct}")

# Scenario 1: Add a new key-value pair
print("\n--- Adding new key 'orange' ---")
update_dictionary(dct, 'orange', 3)


# Scenario 2: Update an existing key
print("\n--- Updating existing key 'apple' ---")
update_dictionary(dct, 'apple', 10)


# Scenario 3: Update another existing key
print("\n--- Updating existing key 'banana' with a new type of value ---")
update_dictionary(dct, 'banana', 'yellow fruit')


# Scenario 4: Add another new key-value pair
print("\n--- Adding new key 'grape' ---")
update_dictionary(dct, 'grape', 5)
```
